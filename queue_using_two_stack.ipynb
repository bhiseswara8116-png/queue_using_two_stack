```python

class QueueUsingTwoStacks:
    def __init__(self):
        self.stack1 = []  # For enqueue operations
        self.stack2 = []  # For dequeue operations

    def enqueue(self, item):
        """Insert an element into the queue."""
        self.stack1.append(item)
        print(f"{item} inserted into the queue.")

    def transfer_elements(self):
        """Transfer elements only when stack2 is empty."""
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())

    def dequeue(self):
        """Remove the front element from the queue."""
        self.transfer_elements()

        if not self.stack2:
            print("Queue is empty. Nothing to remove.")
            return

        item = self.stack2.pop()
        print(f"Removed element: {item}")

    def peek(self):
        """Display the front element without removing it."""
        self.transfer_elements()

        if not self.stack2:
            print("Queue is empty.")
        else:
            print(f"Front element: {self.stack2[-1]}")

    def display(self):
        """Display elements in FIFO order."""
        if not self.stack1 and not self.stack2:
            print("Queue is empty.")
            return

        queue = self.stack2[::-1] + self.stack1
        print("Queue (front to rear):", queue)


def main():
    queue = QueueUsingTwoStacks()

    while True:
        print("\n===== QUEUE USING TWO STACKS =====")
        print("1. Enqueue (Insert)")
        print("2. Dequeue (Remove)")
        print("3. Peek (Front Element)")
        print("4. Display Queue")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ").strip()

        if choice == "1":
            try:
                item = int(input("Enter an integer to insert: "))
                queue.enqueue(item)
            except ValueError:
                print("Invalid input. Please enter an integer.")

        elif choice == "2":
            queue.dequeue()

        elif choice == "3":
            queue.peek()

        elif choice == "4":
            queue.display()

        elif choice == "5":
            print("Thank you for using the program!")
            break

        else:
            print("Invalid choice. Please select from 1 to 5.")


if __name__ == "__main__":
    main()
```
